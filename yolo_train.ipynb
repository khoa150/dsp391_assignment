```python
# import os

# # Tắt telemetry của W&B
# os.environ["WANDB_DISABLE_STATS"] = "true"

import wandb
from ultralytics import YOLO

# Chỉ login, không init project
wandb.login()

# Load pretrained YOLOv8 small
model = YOLO('yolov8x.pt')

# Train/Fine-tune
model.train(
    data='data.yaml', 
    epochs=100,                  
    imgsz=416,                   
    batch=16,
    lr0=0.002,                    
    optimizer='AdamW',            
    patience=15,     
    project='YOLOv8_Finetune',   
    name='exp_2_416_AdamW'        
)

# Kết thúc session W&B
wandb.finish()
```

```python
# ================== CÀI THƯ VIỆN ==================

# ================== IMPORT ==================
import os
import numpy as np
import yaml
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from ultralytics import YOLO

# ================== ĐƯỜNG DẪN ==================
data_yaml = r'data.yaml'
base_dir = os.path.dirname(data_yaml)

with open(data_yaml, 'r') as f:
    data_cfg = yaml.safe_load(f)

test_img_dir = os.path.abspath(os.path.join(base_dir, data_cfg['test']))
test_label_dir = test_img_dir.replace('images', 'labels')

print(f"✅ Test images: {test_img_dir}")
print(f"✅ Test labels: {test_label_dir}")

class_names = data_cfg['names']
num_classes = len(class_names)
print(f"✅ Classes ({num_classes}): {class_names}")

# ================== LOAD MODEL ==================
model = YOLO(r'YOLOv8_Finetune\exp_2_416_AdamW\weights\best.pt')

# ================== LỌC NHÃN HỢP LỆ ==================
label_files = [f for f in os.listdir(test_label_dir) if f.endswith('.txt')]
valid_label_files = []

for label_file in label_files:
    label_path = os.path.join(test_label_dir, label_file)
    if os.path.getsize(label_path) == 0:  # bỏ qua file rỗng
        print(f"⚠️ Bỏ qua file rỗng: {label_file}")
        continue
    valid_label_files.append(label_file)

print(f"\n✅ Tổng số file hợp lệ: {len(valid_label_files)} / {len(label_files)}")

# ================== DỰ ĐOÁN ==================
results = model.predict(
    source=test_img_dir,
    conf=0.25,
    iou=0.5,
    imgsz=416,
    save=False,
    verbose=False
)

# ================== HÀM IOU ==================
def box_iou(box1, box2):
    xA = max(box1[0], box2[0])
    yA = max(box1[1], box2[1])
    xB = min(box1[2], box2[2])
    yB = min(box1[3], box2[3])
    interArea = max(0, xB - xA) * max(0, yB - yA)
    if interArea == 0:
        return 0.0
    box1Area = (box1[2] - box1[0]) * (box1[3] - box1[1])
    box2Area = (box2[2] - box2[0]) * (box2[3] - box2[1])
    return interArea / float(box1Area + box2Area - interArea)

# ================== CHUYỂN VỀ TỌA ĐỘ ABSOLUTE ==================
def yolo_to_xyxy(xc, yc, w, h, img_w, img_h):
    x1 = (xc - w / 2) * img_w
    y1 = (yc - h / 2) * img_h
    x2 = (xc + w / 2) * img_w
    y2 = (yc + h / 2) * img_h
    return [x1, y1, x2, y2]

# ================== TẠO CONFUSION MATRIX ==================
y_true, y_pred = [], []

for i, result in enumerate(results):
    img_path = result.path
    img_name = os.path.basename(img_path)
    label_path = os.path.join(test_label_dir, os.path.splitext(img_name)[0] + '.txt')

    if not os.path.exists(label_path) or os.path.getsize(label_path) == 0:
        continue

    # đọc ảnh và nhãn thật
    img = result.orig_img
    h, w = img.shape[:2]

    gt_data = np.loadtxt(label_path).reshape(-1, 5)
    gt_boxes = [yolo_to_xyxy(*gt[1:], w, h) for gt in gt_data]
    gt_classes = [int(gt[0]) for gt in gt_data]

    pred_boxes = result.boxes.xyxy.cpu().numpy()
    pred_classes = result.boxes.cls.cpu().numpy().astype(int)
    pred_conf = result.boxes.conf.cpu().numpy()

    matched_gt = set()
    for pb, pc, conf in zip(pred_boxes, pred_classes, pred_conf):
        matched = False
        for gi, gb in enumerate(gt_boxes):
            if gi in matched_gt:
                continue
            if box_iou(pb, gb) >= 0.5:
                y_true.append(gt_classes[gi])
                y_pred.append(pc)
                matched_gt.add(gi)
                matched = True
                break
        if not matched:
            y_true.append(num_classes)  # dự đoán sai
            y_pred.append(pc)
    # GT không được dự đoán
    for gi in range(len(gt_boxes)):
        if gi not in matched_gt:
            y_true.append(gt_classes[gi])
            y_pred.append(num_classes)  # miss detection

# ================== VẼ CONFUSION MATRIX ==================
labels_display = class_names + ["background/miss"]
cm = confusion_matrix(y_true, y_pred, labels=list(range(num_classes + 1)))

plt.figure(figsize=(8, 6))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels_display)
disp.plot(cmap="Blues", values_format='d', ax=plt.gca())
plt.title("Confusion Matrix - YOLOv8")
plt.show()
```
